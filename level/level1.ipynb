```python
def lru(pages, capacity):
    frames = []
    page_faults = 0

    for page in pages:
        if page not in frames:
            page_faults += 1
            if len(frames) == capacity:
                # Удаляем самую старую по использованию страницу (первую в списке)
                frames.pop(0)
            frames.append(page)
        else:
            # Перемещаем страницу в конец списка, так как она использовалась только что
            frames.remove(page)
            frames.append(page)

    return page_faults
```

```python
def optimal(pages, capacity):
    frames = []
    page_faults = 0

    for i in range(len(pages)):
        page = pages[i]
        if page not in frames:
            page_faults += 1
            if len(frames) < capacity:
                frames.append(page)
            else:
                # Ищем страницу в кадрах, которая дальше всего в будущем
                farthest = -1
                victim_idx = 0
                for idx, f_page in enumerate(frames):
                    if f_page not in pages[i + 1:]:
                        victim_idx = idx
                        break
                    else:
                        next_use = pages[i + 1:].index(f_page)
                        if next_use > farthest:
                            farthest = next_use
                            victim_idx = idx
                frames[victim_idx] = page

    return page_faults
```

```python
def optimal(refs, n_frames):
    frames = []
    faults = 0
    for i, page in enumerate(refs):
        if page in frames:
            continue
        faults += 1
        if len(frames) < n_frames:
            frames.append(page)
            continue
        future = refs[i + 1 :]

        # Поиск страницы-жертвы, которая понадобится позже всех
        victim = max(
            frames, key=lambda p: future.index(p) if p in future else len(future)
        )

        frames[frames.index(victim)] = page
    return faults
```
